# Traffic Data Analysis & Exploration
This notebook performs exploratory analysis of the smart city traffic datasets to understand vehicle patterns and label distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid')
df = pd.read_csv('../datasets/raw/TrafficTwoMonth.csv')
df.head()
```

## Summary Statistics and Data Quality Check

```python
print("Shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
df.describe()
```

## Distribution of Vehicle Types
We inspect the vehicle composition counts.

```python
vehicles = ['CarCount', 'BikeCount', 'BusCount', 'TruckCount']
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for i, col in enumerate(vehicles):
    ax = axes[i//2, i%2]
    sns.histplot(df[col], kde=True, ax=ax, color='teal')
    ax.set_title(f'Distribution of {col}')
plt.tight_layout()
plt.show()
```

## Congestion Target Classes
Let's look at the class distribution of `Traffic Situation`.

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Traffic Situation', data=df, order=['low', 'normal', 'high', 'heavy'], palette='viridis')
plt.title('Traffic Situation Frequency')
plt.show()
```
